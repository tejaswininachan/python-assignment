# Data Types in Python

# int -> Whole numbers
a = 10

# float -> Decimal numbers
b = 10.5

# str -> Text/String
c = "Tejaswini"

# bool -> True or False
d = True

# list -> Ordered and mutable collection
e = [1, 2, 3]

# tuple -> Ordered and immutable collection
f = (1, 2, 3)

# dict -> Key-value pairs
g = {"name": "Tejaswini", "age": 20}

# set -> Unique unordered values
h = {1, 2, 3}

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
print(type(g))
print(type(h))
